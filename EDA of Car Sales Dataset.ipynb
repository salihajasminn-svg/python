```python
#matplotlib
```

```python

Carsale['Company'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Car Companies by Sales")
plt.show()
```

```python
#Bar chart
```

```python
plt.figure(figsize=(8,5))
Carsale['Company'].value_counts().head(10).plot(kind='bar', color='skyblue')
plt.title("Top 10 Car Companies by Sales Count")
plt.xlabel("Company")
plt.ylabel("Number of Cars Sold")
plt.show()
```

```python
#Line chart
```

```python
Carsale['Year'] = Carsale['Date'].dt.year
yearly_sales =Carsale.groupby('Year')['Price ($)'].sum()

plt.figure(figsize=(6,4))
plt.plot(yearly_sales.index, yearly_sales.values, marker='o')
plt.title("Total Sales Over Years")
plt.xlabel("Year")
plt.ylabel("Total Sales ($)")
plt.show()
```

```python
#Pie chart
```

```python
plt.figure(figsize=(6,6))
Carsale['Transmission'].value_counts().plot(
    kind='pie', autopct='%1.1f%%', startangle=90
)
plt.title("Transmission Type Distribution")
plt.ylabel("")
plt.show()
```

```python
#Seaborn
```

```python
#Box plot
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(x='Body Style', y='Price ($)', data=Carsale)
plt.title("Price Distribution by Body Style")
plt.xticks(rotation=45)
plt.show()
```

```python
#Histogram
```

```python
plt.figure(figsize=(8,5))
sns.histplot(Carsale['Price ($)'], bins=30, kde=True)
plt.title("Car Price Distribution")
plt.xlabel("Price ($)")
plt.ylabel("Frequency")
plt.show()
```

```python
#Scatter plot
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='Annual Income', y='Price ($)', data=Carsale, alpha=0.5)
plt.title("Annual Income vs Car Price")
plt.xlabel("Annual Income")
plt.ylabel("Car Price ($)")
plt.show()
```
