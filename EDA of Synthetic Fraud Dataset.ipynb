```python
#Bar Chart (Transaction Type Count)

plt.figure(figsize=(8,5))
sns.countplot(data=synthetic_fraud_dataset, x='transaction_type')
plt.title("Transaction Type Distribution")
plt.xticks(rotation=45)
plt.show()
```

```python
#Line Chart (Transactions by Hour)

hourly_txn = synthetic_fraud_dataset.groupby('hour')['amount'].sum()

plt.figure(figsize=(10,5))
plt.plot(hourly_txn.index, hourly_txn.values, marker='o')
plt.title("Total Transaction Amount by Hour")
plt.xlabel("Hour")
plt.ylabel("Amount")
plt.grid(True)
plt.show()
```

```python
#Pie Chart (Fraud vs Non-Fraud)

fraud_counts = synthetic_fraud_dataset['is_fraud'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    fraud_counts,
    labels=['Not Fraud', 'Fraud'],
    autopct='%1.1f%%',
    startangle=90
)
plt.title("Fraud Distribution")
plt.show()
```

```python
#Box Plot (Amount vs Fraud)

plt.figure(figsize=(8,5))
sns.boxplot(data=synthetic_fraud_dataset, x='is_fraud', y='amount')
plt.title("Transaction Amount by Fraud Status")
plt.show()
```

```python
#Histogram (Transaction Amount)

plt.figure(figsize=(8,5))
sns.histplot(synthetic_fraud_dataset['amount'], bins=30, kde=True)
plt.title("Transaction Amount Distribution")
plt.xlabel("Amount")
plt.show()
```

```python
#Scatter Plot (Amount vs Device Risk Score)

plt.figure(figsize=(8,5))
sns.scatterplot(
    data=synthetic_fraud_dataset,
    x='device_risk_score',
    y='amount',
    hue='is_fraud'
)
plt.title("Amount vs Device Risk Score")
plt.show()
```

```python
#Correlation Heatmap (Bonus)

plt.figure(figsize=(8,6))

numeric_cols = synthetic_fraud_dataset.select_dtypes(include=np.number)

sns.heatmap(
    numeric_cols.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Matrix")
plt.show()
```
